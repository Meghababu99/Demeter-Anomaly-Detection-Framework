```python
import pandas as pd
import psycopg2
from sqlalchemy import create_engine
from pathlib import Path

# Define the station object
class Station:
    def __init__(self, code):
        self.code = code

# Define the flatten_dict function
def flatten_dict(d: dict, parent_key: str = '', sep: str = '_') -> dict:
    """Recursively flatten a nested dictionary."""
    items = []
    for k, v in d.items():
        new_key = f"{parent_key}{sep}{k}" if parent_key else k
        if isinstance(v, dict):
            items.extend(flatten_dict(v, new_key, sep=sep).items())
        else:
            items.append((new_key, v))
    return dict(items)

# Initialize the SAODataset
station = Station(code="AT138")
# input_file_path = "C:\\NOA\\Data\\SAO\\AT138_20080101(001).SAO"
input_file_path ="C:\\NOA\\Data\\ATHENS\\AT138_20050101(001).SAO"
dataset = SAODataset(
    path=input_file_path,  
    start="2005-01-01T00:00:00",  
    station=station  
)
# code="AT138"
# station = Station(code="AT138")
# dataset = SAODataset(
#     path="C:\\NOA\\Ionograms\\AT138_20240321(081).SAO",  # Path to the SAO file
#     start="2024-03-21T08:00:00",                         # Start time
#     station=station                                      # Station object
# )
# Collect all SAO objects into a list
sao_objects = [sao for sao in dataset.read()]

# Flatten each SAO object and store in a list
flattened_data = [flatten_dict(sao.dict()) for sao in sao_objects]

# Create a DataFrame from the flattened data
df = pd.DataFrame(flattened_data)

# Database connection details
DB_NAME = "saodb"  # The name of your database
DB_USER = "megha"  # The user that can access the database
DB_PASSWORD = "meghababu"  # Password for the user
DB_HOST = "localhost"  # Host, usually localhost for local setups
DB_PORT = "5433"  # Port for PostgreSQL

# Create a connection engine using SQLAlchemy
try:
    # Define the connection string
    connection_string = f'postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}'
    
    # Create the engine
    engine = create_engine(connection_string)

    # Define the table name
    table_name = f'AT138_2005'

    # Save DataFrame to PostgreSQL
    df.to_sql(table_name, engine, if_exists="append", index=False)

    print(f"Data stored in PostgreSQL table '{table_name}' successfully.")

except Exception as e:
    print(f"An error occurred: {e}")
```
