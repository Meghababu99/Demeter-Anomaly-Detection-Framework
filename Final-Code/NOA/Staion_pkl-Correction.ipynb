```python
import pandas as pd
from datetime import datetime

file_path = r"C:\NOA\Data\Anomaly-station\SAA0K_20100101(001).SAO"

timestamps = []

with open(file_path, 'r') as file:
    for line in file:
        if line.startswith("FF"):
            try:
                year = line[2:6]
                doy = line[6:9]  # Not used now
                month = line[9:11]
                day = line[11:13]
                hour = line[13:15]
                minute = line[15:17]
                second = line[17:19]

                timestamp_str = f"{year}-{month}-{day} {hour}:{minute}:{second}"
                timestamp = datetime.strptime(timestamp_str, "%Y-%m-%d %H:%M:%S")
                
                timestamps.append(timestamp.strftime("%Y-%m-%d %H:%M:%S"))

            except Exception as e:
                print(f"Error parsing line: {line.strip()}")
                print(f"Exception: {e}")

# Create DataFrame
df = pd.DataFrame(timestamps, columns=["sounding_timestamp"])
print(df)
```
